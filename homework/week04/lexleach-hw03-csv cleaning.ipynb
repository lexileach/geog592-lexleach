## Cleaning CSV Data (NC Hazardous Waste Sites)

## **Part 1:** Read the original CSV file

```python
import csv

filename = "nc_hazardous_sites.csv"

with open(filename, "r") as file:
    reader = csv.DictReader(file)
    rows = list(reader)

print("Records found:", len(rows))
```

## **Part 2:** Clean the CSV file using Python’s built-in csv module

```python
print("Number of rows:", len(rows))
print("Number of columns:", len(reader.fieldnames))
print("Column names:")
print(reader.fieldnames)
```

```python
for row in rows[:3]:
    print(row)
```

```python
selected_columns = []

for row in rows:

    selected_columns.append({
        "EPAID": row["EPAID"],
        "SITENAME": row["SITENAME"],
        "SITEADDR": row["SITEADDR"],
        "SITECITY": row["SITECITY"],
        "SITECOUNTY": row["SITECOUNTY"],
        "LATITUDE": row["LATITUDE"],
        "LONGITUDE": row["LONGITUDE"]
    })

for row in selected_columns[:5]:
    print(row)
```

```python
def can_be_float(value):
    try:
        float(value)
        return True
    except ValueError:
        return False
```

```python
error_found = False
missing_count = 0

for row in selected_columns:
    value = row["LATITUDE"]

    if value != "" and not can_be_float(value):
        print(
            row["SITENAME"],
            "has suspicious LATITUDE:",
            value
        )
        error_found = True

    if value == "":
        missing_count += 1
        print(
            row["SITENAME"],
            "has an empty LATITUDE"
        )

if not error_found:
    print("All LATITUDE entries have a consistent numeric format.")

print(
    "Number of empty LATITUDE entries:",
    missing_count
)
```

```python
error_found = False
missing_count = 0

for row in selected_columns:
    value = row["LONGITUDE"]

    if value != "" and not can_be_float(value):
        print(
            row["SITENAME"],
            "has suspicious LONGITUDE:",
            value
        )
        error_found = True

    if value == "":
        missing_count += 1
        print(
            row["SITENAME"],
            "has an empty LONGITUDE"
        )

if not error_found:
    print("All LONGITUDE entries have a consistent numeric format.")

print(
    "Number of empty LONGITUDE entries:",
    missing_count
)
```

```python
for row in selected_columns:
    value = row["LATITUDE"]

    if value != "" and can_be_float(value):
        value = float(value)

        if value < 33 or value > 37:
            print(
                row["SITENAME"],
                "has suspicious latitude:",
                value
            )
```

```python
for row in selected_columns:
    value = row["LONGITUDE"]

    if value != "" and can_be_float(value):
        value = float(value)

        if value < -85 or value > -75:
            print(
                row["SITENAME"],
                "has suspicious longitude:",
                value
            )
```

```python
missing_count = 0

for row in selected_columns:
    value = row["SITENAME"]

    if value == "":
        missing_count += 1
        print("A record is missing a SITENAME")

if missing_count == 0:
    print("All records have a SITENAME.")
else:
    print(
        "Number of records missing SITENAME:",
        missing_count
    )
```

```python
missing_count = 0

for row in selected_columns:
    value = row["SITECOUNTY"]

    if value == "":
        missing_count += 1
        print(
            row["SITENAME"],
            "is missing a SITECOUNTY"
        )

if missing_count == 0:
    print("All records have a SITECOUNTY.")
else:
    print(
        "Number of records missing SITECOUNTY:",
        missing_count
    )
```

```python
import re

for row in selected_columns:
    value = row["SITECOUNTY"]

    if re.search(r"[^A-Za-z0-9 ]", value):
        print(
            row["SITENAME"],
            "has a SITECOUNTY with a special character:",
            value
        )
```

```python
for row in selected_columns:
    print(row["SITECOUNTY"])
```

```python
fieldnames = selected_columns[0].keys()

with open("nc_hazardous_sites_clean.csv", "w", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(selected_columns)
```
