## Cleaning Açaí Production Data for Pará



## **Part 1:** Read the original CSV file

```python
import csv

filename = "tabela1613.csv"

with open(filename, "r", encoding='utf-8') as file:
    reader = csv.reader(file)

    for x in range(4):
        next(reader)

    reader = csv.DictReader(file)
    rows = list(reader)

print("Records found:", len(rows))
```

## **Part 2:** Inspect the data and identify the columns needed for this assignment

```python
print("Number of rows:", len(rows))
print("Number of columns:", len(reader.fieldnames))
print("Column names:")
print(reader.fieldnames)

for row in rows[:6]:
    print(row)
```

## **Part 3:** Filter the data so that only municipalities in Pará (PA) remain

```python
# Select the columns that are relevant
selected_columns = []

for row in rows:

    selected_columns.append({
        "LocationCode": row["Cód."],
        "Municipality": row["Município"],
        "Acai": row["Açaí"]
    })

for row in selected_columns[:5]:
    print(row)

# Print first row
selected_columns[0]

# Select rows with (PA)
selected_rows = []

for row in selected_columns:
    value = row["Municipality"]

    if value is not None and '(PA)' in value:
            selected_rows.append(row)

print(selected_rows)
print(len(selected_rows))
```

## **Part 4:** Keep only the columns containing:
- the municipality code (Cod / Cód.), which will later be used to join the CSV to a GIS layer;
- the quantity of açaí produced.

```python
# Select final columns of interest
final_col = []

for row in selected_rows:

    final_col.append({
        "LocationCode": row["LocationCode"],
        "Acai": row["Acai"]
    })

for row in final_col[:5]:
    print(row)
```

# **Part 5:** Check the selected data for problems such as: 
- missing values
- non-numeric values in the açaí production column
- unnecessary rows or columns
- data type

```python
# Check for missing values
nv = []

for row in final_col:
    try:
        float(row['Acai'])
    except ValueError:
        nv.append(row['Acai'])

print(nv)

# Define integer function
def can_be_int(value):
    try:
        int(value)
        return True
    except ValueError:
        return False

# Run an error check to identify which rows have errors
error_found = False

for row in final_col:
    value = row["Acai"]

    if value == "-":
        print(
            row["LocationCode"],
            "has an 'Acai' value of 0"
        )
        error_found = True

    elif value == "...":
        print(
            row["LocationCode"],
            "has a missing 'Acai' value"
        )

    elif not can_be_int(value):
        print(
            row["LocationName"],
            "has a suspicious Acai value:",
            value
        )
        error_found = True

if not error_found:
    print("Everything looks good.")
```

# **Part 6:** Make any necessary cleaning decisions and show the Python code you used

```python
# Loop for replacing '-' with 0
for row in final_col:
    value = row["Acai"]

    if value == "-":
        row["Acai"] = "0"

    if value == "...":
        row["Acai"] = "NA"

print(final_col[:6])
print(final_col[3])
```

# **Part 7:** Export the cleaned data as a new CSV file

```python
# Export as CSV
fieldnames = final_col[0].keys() ## this gets the names of the columns (or keys)

with open("clean_acai_data.csv", "w", newline="") as file: 
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(final_col)
```
